## <a class="anchor" id="2_Introduction"></a>  2. Introduction 

[Back to Table of Contents](#Table_of_Contents) 

According to a plan now suspended and under budget review, a Sample Retrieval Lander will be sent to Mars together with a NASA-led Mars rocket and two miniature Mars helicopters. In Jezero Crater, the lander will touchdown near the location of the Perseverance landing. The Perseverance rover will be the main vehicle for delivering samples to the Sample Retrieval Lander, handing off a collection of sample tubes carried on board, having gathered and stored a wide range of materials. The tubes will be moved into a sample container on the lander's rocket by a robotic arm supplied by the ESA. A supplementary capability will be provided by two helicopters, capable of picking up extra samples that Perseverance had hidden on the surface and return them to the lander for loading aboard the Mars rocket. The container of sample tubes will be carried into orbit by the Mars Ascent Vehicle, the first rocket to ever launch from the surface of Mars. The NASA-supplied Capture, Containment, and Return System and Earth Entry System would be mounted aboard the ESA Earth Return Orbiter. The samples would be captured and contained by the orbiter's return system, which would then deposit them in the Earth Entry System. The entrance vehicle would subsequently be transported back to Earth's vicinity by the Earth Return Orbiter, where it would detach and land safely.

 ## <a class="anchor" id="Staging"></a> Staging

[Back to Table of Contents](#Table_of_Contents) 

Staging

The Mars Ascent Vehicle employs a two-stage configuration to optimize mass efficiency and ascent performance under Martian gravity.

First Stage
- Provides the initial thrust required for lift-off from the Martian surface.
- Overcomes Mars gravity and atmospheric drag.
- Operates for a short, high-thrust burn period.
- Is jettisoned after propellant depletion to reduce dead mass.

Second Stage
- Ignites immediately after first-stage separation.
- Responsible for *achieving orbital velocity* and *orbit insertion*.
- Carries the payload (scientific samples) into Mars orbit.
- Optimized for higher efficiency and precise trajectory control.

This staging strategy minimizes total vehicle mass while maximizing ascent efficiency and mission reliability.


## <a class="anchor" id="Aerodynamic_stress"></a> Aerodynamic stress

[Back to Table of Contents](#Table_of_Contents) 

During ascent through the Martian atmosphere, the Mars Ascent Vehicle (MAV) is subjected to aerodynamic stress due to the relative motion of air and vehicle surfaces.  
The primary aerodynamic forces are drag and dynamic pressure, which must be carefully considered to prevent structural failure.

 1. Dynamic Pressure

The **dynamic pressure** $q$ is defined as:

$$
q = \frac{1}{2} \rho v^2
$$

where:  
- $\rho$ = atmospheric density on Mars [kg/m³]  
- $v$ = velocity of the vehicle relative to the atmosphere [m/s]

2. Drag Force

The **drag force** $F_D$ acting on the vehicle is:

$$
F_D = C_D \, q \, A_{\text{ref}}
$$

where:  
- $C_D$ = drag coefficient (depends on vehicle shape)  
- $A_{\text{ref}}$ = reference area of the MAV [m²]  

3. Structural Stress

The **aerodynamic stress** experienced by the MAV can be approximated as:

$$
\sigma = \frac{F_D}{A_{\text{struct}}}
$$

where $A_{\text{struct}}$ is the cross-sectional area of the structural components experiencing the load.

## Additional questions

**Q-A1** How would you locate and draw the main axis of the orbit and the Argument of the Periapsis?

The main (major) axis of an orbit is defined by the line connecting the periapsis
and apoapsis of the trajectory. Its orientation in inertial space is determined
by the argument of periapsis, ω, measured from a reference direction.

For a planar two-body orbit, the specific angular momentum is:
$$
\mathbf{l} = \mathbf{r} \times \mathbf{v}
$$
The eccentricity vector is given by:

$$
\mathbf{e} = \frac{1}{k}\left( \mathbf{v} \times \mathbf{l} \right)
$$

The direction of the eccentricity vector points toward periapsis. Therefore,
the argument of periapsis is obtained from:

$$
\omega = \tan^{-1}\left(\frac{e_y}{e_x}\right)
$$

```
(* Position and velocity at orbit injection *)
rVec = {rSolSC[tfin] Cos[ThetaAngSolSC[tfin]],rSolSC[tfin] Sin[ThetaAngSolSC[tfin]],0};

vVec = { rSCdotSol[tfin] Cos[ThetaAngSolSC[tfin]] - vtSolSC[tfin] Sin[ThetaAngSolSC[tfin]],
  rSCdotSol[tfin] Sin[ThetaAngSolSC[tfin]] + vtSolSC[tfin] Cos[ThetaAngSolSC[tfin]],0};

(* Specific angular momentum *)
lVec = Cross[rVec, vVec];

(* Eccentricity vector *)
eVec = (Cross[vVec, lVec]/kMars) - (rVec/Norm[rVec]);

(* Eccentricity magnitude *)
eMag = Norm[eVec]

(* Argument of periapsis *)
omegaAOP = ArcTan[eVec[[1]], eVec[[2]]] * 180/Pi

(* Periapsis and apoapsis vectors *)
rpVec = SMAAM (1 - EccentricityAM) Normalize[eVec];
raVec = -SMAAM (1 + EccentricityAM) Normalize[eVec];

MajorAxisPlot =Graphics[{ Black, Thick, Line[{{rpVec[[1]], rpVec[[2]]}, {raVec[[1]], raVec[[2]]}}]}];

Show[PLOTMarsAll, PLOTSCTraj1Orb, MajorAxisPlot,
 PlotRange -> {{-15000 10^3, 5000 10^3}, {-10000 10^3, 10000 10^3}}]
```

**Q-A3** How are your conclusions affected by staging? (assume two stages, initially equal. Neglect atmospheric drag). 

Staging increases performance by discarding inert mass, allowing the remaining
propellant to accelerate a smaller vehicle mass. For a fixed total launch mass
and engine performance, staging increases the achievable velocity increment
and alters the injection conditions into orbit.

In this study, a two-stage Mars Ascent Vehicle is modeled with:
- equal stage masses,
- instantaneous mass drop at staging,
- no atmospheric drag.

The final orbit is compared to the single-stage case by examining the periapsis
altitude and orbital energy.

**Q-A4** How are your conclusions affected by air drag? (assume one stage). 

Atmospheric drag opposes the vehicle velocity and reduces the net acceleration
during ascent. Although the Martian atmosphere is thin, drag losses can be
significant during the early phase of flight when the vehicle velocity and
atmospheric density are highest.

In this analysis, the same single-stage Mars Ascent Vehicle is simulated with
and without aerodynamic drag. The effect of drag is quantified by comparing the
resulting periapsis altitude and orbital energy after burnout.

```
kMars = 4.282837*10^13; 
RadiusMars = 3.3895*10^6; 
MassRocket0 = 500.; 
DryMass = 75;  
FThrustLAS = 8388; 
ISP = 300; 
gMars = 3.728; 
gEarth = 9.8065;

(* Exhaust velocity and mass flow rate *)
vexh = ISP gEarth;      
mdot = FThrustLAS / vexh;

(* Rocket mass as a function of time, capped at dry mass *)
MassRocket[t_?NumericQ] := Max[MassRocket0 - mdot t, DryMass];

(* Atmospheric model *)
CD = 0.2; 
rRocket = 0.3; 
Aeff = Pi rRocket^2;

kB = 1.380649*10^-23; 
mu = 1.66053906892*10^-27;
H0 = (kB/ mu) 210 / (43.34 gMars); 

rhoAtm[r_?NumericQ] := 0.02 Exp[-(r - RadiusMars)/H0];

(* Thrust angle function *)
ThrustPhi[t_?NumericQ, phi0_?NumericQ, Tprogr_?NumericQ] := ArcTan[(1 - t/Tprogr) Tan[phi0 Degree]];

(* Drag force function *)
Fdrag[r_?NumericQ, rdot_?NumericQ, includeDrag_: True] := If[includeDrag, -0.5 CD Aeff rhoAtm[r] rdot Abs[rdot], 0];

(* Function to compute periapsis with drag *)
ComputePeriapsisWithDrag[phi0_?NumericQ, includeDrag_: True] := Module[
  {tfin, r0, rdot0, theta0, thetadot0, sol, rEnd, vtEnd, lEnd, ESpec, SMA, ecc, rp, ra},
  
  r0 = RadiusMars; rdot0 = 0; theta0 = 0; thetadot0 = 0;
  tfin = (MassRocket0 - DryMass)/mdot;  (* burn until dry mass reached *)
  
  sol = NDSolve[
    { rSC''[t] == rSC[t] (ThetaAngSC'[t]^2) - kMars/(rSC[t]^2) +(FThrustLAS / MassRocket[t]) Sin[ThrustPhi[t, phi0, tfin]] + Fdrag[rSC[t], rSC'[t], includeDrag]/MassRocket[t],
      ThetaAngSC''[t] == -(2/rSC[t]) rSC'[t] ThetaAngSC'[t] +(1/rSC[t]) (FThrustLAS / MassRocket[t]) Cos[ThrustPhi[t, phi0, tfin]],
      rSC[0] == r0, rSC'[0] == rdot0,
      ThetaAngSC[0] == theta0, ThetaAngSC'[0] == thetadot0},
    {rSC, ThetaAngSC}, {t, 0, tfin},
    AccuracyGoal -> 13, PrecisionGoal -> 13, MaxSteps -> Infinity,
    Method -> {"StiffnessSwitching", "NonstiffTest" -> False} ];
  
  (* Final orbital parameters *)
  rEnd = rSC[tfin] /. sol[[1]];
  vtEnd = (ThetaAngSC'[tfin] rSC[tfin]) /. sol[[1]];
  lEnd = rEnd*vtEnd;
  ESpec = 0.5*((rSC'[tfin]^2 + (ThetaAngSC'[tfin] rSC[tfin])^2) /. sol[[1]]) - kMars/rEnd;
  SMA = -kMars/(2 ESpec);
  ecc = Sqrt[1 + 2 ESpec lEnd^2/kMars^2];
  rp = SMA*(1 - ecc);
  ra = SMA*(1 + ecc);
  
  {rp, ra}];

(* Sweep over thrust angles *)
phi0List = Range[-40, 50, 1];
orbitDataNoDrag = ComputePeriapsisWithDrag[#, False] & /@ phi0List;
orbitDataDrag   = ComputePeriapsisWithDrag[#, True] & /@ phi0List;

rpNoDrag = orbitDataNoDrag[[All, 1]];
rpDrag   = orbitDataDrag[[All, 1]];

safeNoDrag = Pick[phi0List, Map[# > RadiusMars &, rpNoDrag]];
safeDrag   = Pick[phi0List, Map[# > RadiusMars &, rpDrag]];

Print["Safe φ₀ range without drag: ", safeNoDrag];
Print["Safe φ₀ range with drag:    ", safeDrag];

(* Plot periapsis altitude vs φ₀ *)
bbb = ListLinePlot[{ Transpose[{phi0List, rpNoDrag - RadiusMars}], Transpose[{phi0List, rpDrag - RadiusMars}]},PlotStyle -> {Yellow, Blue},
  AxesLabel -> {"φ₀ [deg]", "Periapsis altitude [m]"},PlotRange -> {{-40, 50}, {-100*10^3, 20*10^3}},
  PlotLabel -> "Periapsis vs Initial Thrust Angle φ₀: Drag vs No Drag",
  GridLines -> Automatic]
```

**Q-A5** How are you conclusions altered in the full case of staging with unequal stages and atmospheric drag?  

```
gMars = 3.728;
kMars = 4.282837*10^13;
RadiusMars = 3.3895*10^6;
CD = 0.2;
rRocket = 0.3;
Aeff = Pi rRocket^2;


kB = 1.380649*10^-23;
mu = 1.66053906892*10^-27;
H0 = (kB/mu) * 210 / (43.34 * gMars);
rhoAtm[r_] := 0.02 Exp[-(r - RadiusMars)/H0];

(* Thrust angle function *)
ThrustPhi[t_, phi0_] := ArcTan[(1 - t/tfin) Tan[phi0 Degree]];

(* Stage parameters *)
stages = {
  <|"Mass0" -> 300., "FuelMass" -> 200., "ISP" -> 300, "Thrust" -> 8388|>,
  <|"Mass0" -> 150., "FuelMass" -> 100., "ISP" -> 300, "Thrust" -> 5000|>};

(* Mass flow per stage *)
mdott[k_Integer] := stages[[k]]["Thrust"]/(stages[[k]]["ISP"] * gEarth);
BurnOutTime[k_Integer] := stages[[k]]["FuelMass"]/mdott[k];
MassRocket[k_Integer, t_] := Max[stages[[k]]["Mass0"] - mdott[k] t, stages[[k]]["Mass0"] - stages[[k]]["FuelMass"]];

(* Stage index for time t *)
currStage[t_?NumericQ] := Clip[Floor[1 + Count[Accumulate[BurnOutTime /@ Range[Length[stages]]], _?(# <= t &)] - 1], {1, Length[stages]}];

(* Function to compute periapsis altitude *)
periapsisAltitudeStage[phi_?NumericQ] := Module[
  {sol, rFunc, thetaFunc, rdot, vt, energy, l, e, a, rp, stageTimes, tfin, currStageIndex},

  (* Stage timing *)
  stageTimes = Prepend[Accumulate[BurnOutTime /@ Range[Length[stages]]], 0];
  tfin = Last[stageTimes];

  (* Define the stage based on the current time t *)
  currStageIndex = currStage[tfin]; (* This ensures we get the correct stage *)

  sol = NDSolve[{rSC''[t] == rSC[t] (ThetaAngSC'[t]^2) - kMars/rSC[t]^2 +(stages[[currStageIndex]]["Thrust"]/MassRocket[currStageIndex, t]) Sin[ThrustPhi[t, phi]],
      ThetaAngSC''[t] == -(2/rSC[t]) rSC'[t] ThetaAngSC'[t] + (1/rSC[t]) (stages[[currStageIndex]]["Thrust"]/MassRocket[currStageIndex, t]) Cos[ThrustPhi[t, phi]],
      rSC[0] == RadiusMars, rSC'[0] == 0, 
      ThetaAngSC[0] == 0, ThetaAngSC'[0] == 0},
    {rSC, ThetaAngSC}, {t, 0, tfin},MaxSteps -> Infinity][[1]];

  rFunc[t_] := Evaluate[rSC[t] /. sol];
  thetaFunc[t_] := Evaluate[ThetaAngSC[t] /. sol];

  rdot = rFunc'[tfin];
  vt = rFunc[tfin] thetaFunc'[tfin];

  energy = (rdot^2 + vt^2)/2 - kMars/rFunc[tfin];
  l = rFunc[tfin]*vt;

  e = Sqrt[1 + (2 energy l^2)/kMars^2];
  a = -kMars/(2 energy);
  rp = a*(1 - e);

  rp - RadiusMars];

(* Sweep phi0 and compute periapsis altitude *)
phiList = Range[-40, 50, 1];
dataStage = Table[{phi, periapsisAltitudeStage[phi]}, {phi, phiList}];

(* Maximum periapsis and corresponding phi0 *)
maxPointStage = MaximalBy[dataStage, Last][[1]];
phiMaxStage = maxPointStage[[1]];
rpMaxStage = maxPointStage[[2]];

(* Plot *)
periapsisPlott = ListPlot[ dataStage,AxesLabel -> {"φ₀ (degrees)", "Periapsis altitude (m)"},Joined -> True,
  PlotStyle -> Magenta,GridLines -> Automatic,AspectRatio -> 1];

(* Show final plot *)
Show[periapsisPlott, PlotRange -> {{-40, 50}, {-100*10^3, 20*10^3}}, AspectRatio -> 1,
  Epilog -> {Red, PointSize[0.02], Point[{phiMaxStage, rpMaxStage}],Text[Style[Row[{"φ₀ = ", NumberForm[phiMaxStage, {3, 1}], "°\n",
             "Periapsis = ", ScientificForm[rpMaxStage, 3], " m"}], 12, Red],
      {phiMaxStage + 1, rpMaxStage}] }, PlotLabel -> "Mars Ascent Vehicle: Periapsis vs Initial Thrust Angle φ₀"]
```
